```python
import gensim
import pandas as pd
```

```python
df=pd.read_json("/kaggle/input/datasets/divyapriya455/reviews-cell-phones-and-accessories/Cell_Phones_and_Accessories_5.json",lines=True)
df.head(10)
```

```python
df.shape
```

```python
reviewtxt=df.reviewText.apply(gensim.utils.simple_preprocess)
reviewtxt
```

```python
reviewtxt[0]
```

```python
model=gensim.models.Word2Vec(window=10,min_count=2,workers=4)
```

```python
model.build_vocab(reviewtxt,progress_per=1000)
```

```python
model.train(reviewtxt, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
model.save("./word2vec-amazon-cell-accessories-reviews-short.model")
```

```python
model.wv.most_similar("bad")
```

```python
model.wv.similarity(w1="good", w2="nice")
```
